# 3.2. Свойства и информация об элементах

## Truss.nodes

`property nodes`

Возвращает узлы фермы и их позиции.

## Truss.node_labels

`property node_labels`

Возвращает метки узлов фермы.

## Truss.node_positions

`property node_positions`

Возвращает позиции узлов фермы.

## Truss.members

`property members`

Возвращает стержни фермы и их начальные и конечные узлы.

## Truss.member_lengths

`property member_lengths`

Возвращает длины всех стержней фермы.
