## Truss.draw()

`draw(subs_dict=None)`

Возвращает объект графика фермы со всеми её узлами, стержнями, опорами и нагрузками.

> [!note]
> Пользователь должен быть внимателен при вводе значений нагрузок с учётом их направлений. Функция построения использует правило знаков, принятое для отображения нагрузок.
Для правой системы координат с осями X, Y, Z принимается, что опоры таковы, что реакции шарнирно-неподвижной опоры направлены вдоль осей +X и +Y, а реакции шарнирно-подвижной опоры — вдоль оси +Y. Для нагрузки диапазон вводимых углов достигает 360 градусов; на графике этот угол соответствует углу между вектором нагрузки и положительным направлением оси X, отсчитываемым против часовой стрелки.
Например, для угла 90 градусов нагрузка будет направлена вертикально вдоль +Y, а угол 270 градусов также обозначает вертикальную нагрузку, но направленную вдоль −Y.

### Пример

```python
from sympy.physics.continuum_mechanics.truss import Truss
import math
t = Truss()
t.add_node(("A", -4, 0), ("B", 0, 0), ("C", 4, 0), ("D", 8, 0))
t.add_node(("E", 6, 2/math.sqrt(3)))
t.add_node(("F", 2, 2*math.sqrt(3)))
t.add_node(("G", -2, 2/math.sqrt(3)))
t.add_member(("AB","A","B"), ("BC","B","C"), ("CD","C","D"))
t.add_member(("AG","A","G"), ("GB","G","B"), ("GF","G","F"))
t.add_member(("BF","B","F"), ("FC","F","C"), ("CE","C","E"))
t.add_member(("FE","F","E"), ("DE","D","E"))
t.apply_support(("A","pinned"), ("D","roller"))
t.apply_load(("G", 3, 90), ("E", 3, 90), ("F", 2, 90))
t.draw().show()
```
