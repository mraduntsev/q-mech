# 1.2. Свойства сечения и материала

### Beam.area

`property area`

Площадь поперечного сечения балки.

### Beam.cross_section

`property cross_section`

Поперечное сечение балки.

### Beam.elastic_modulus

`property elastic_modulus`

Модуль упругости (модуль Юнга) балки.

### Beam.length

`property length`

Длина балки.

### Beam.reaction_loads

`property reaction_loads`

Возвращает опорные реакции в виде словаря.

### Beam.second_moment

`property second_moment`

Момент инерции (второй момент площади) балки.

### Beam.variable

`property variable`

Символ, который может использоваться в качестве переменной по длине балки для представления распределения нагрузки, эпюр поперечных сил, изгибающих моментов, углов наклона и прогибов. По умолчанию задан как `Symbol('x')`, но это свойство можно изменить.

```python
from sympy.physics.continuum_mechanics.beam import Beam
from sympy import symbols
E, I, A = symbols('E, I, A')
x, y, z = symbols('x, y, z')
b = Beam(4, E, I)
print(b.variable)
b.variable = y
print(b.variable)
b = Beam(4, E, I, A, z)
print(b.variable)
```
