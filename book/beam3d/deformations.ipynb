# 2.4. Расчет перемещений

### Beam3D.slope()

`slope()`
Возвращает список из трех элементов, представляющих угол наклона (первую производную) кривой прогибов вдоль всех трех осей.

### Beam3D.deflection()

`deflection()`
Возвращает список из трех элементов, представляющий кривую прогибов вдоль всех трех осей.

### Beam3D.angular_deflection()

`angular_deflection()`
Возвращает функцию от x, показывающую, как изменяется угол закручивания (угловое перемещение) вдоль оси x балки под действием крутящих моментов.
