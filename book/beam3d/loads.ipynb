# 2.1. Приложение нагрузок

### Beam3D.apply_load()

`apply_load(value, start, order, dir='y')`

Этот метод добавляет силовую нагрузку к конкретному объекту балки.

**Параметры:**

**value** : Sympifyable
> Величина прикладываемой нагрузки.

**dir** : String
> Ось, вдоль которой прикладывается нагрузка.

**order** : Integer
> Порядок прикладываемой нагрузки:
> - Для сосредоточенных сил order = -1
> - Для постоянной распределенной нагрузки order = 0
> - Для линейно изменяющейся (треугольной) нагрузки order = 1
> - Для параболически изменяющейся нагрузки order = 2
> - … и так далее.

### Beam3D.apply_moment_load()

`apply_moment_load(value, start, order, dir='y')`

Этот метод добавляет моментные нагрузки к конкретному объекту балки.

**Параметры:**

**value** : Sympifyable
> Величина прикладываемого момента.

**dir** : String
> Ось, вдоль которой прикладывается момент.

**order** : Integer
> Порядок прикладываемой нагрузки:
> - Для сосредоточенных моментов order = -2
> - Для постоянного распределенного момента order = -1
> - Для линейно изменяющегося (треугольного) момента order = 0
> - Для параболически изменяющегося момента order = 1
> - … и так далее.
